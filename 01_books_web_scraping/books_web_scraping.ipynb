# Webscraping using BeautifulSoup

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
Book_info=[]
url="https://books.toscrape.com/"
r=requests.get(url)
r
```

```python
soup=BeautifulSoup(r.content,'html.parser')
print(soup.prettify())
```

```python
book=soup.find_all('li',class_="col-xs-6 col-sm-4 col-md-3 col-lg-3")
book[:1]
```

```python
for item in book:
  name_tag=item.find('h3').find('a')
  book_name=name_tag.text
  book_link=name_tag['href']
  price=item.find('p',class_="price_color").text
  stock=item.find('p',class_="instock availability").text.strip()
  rating=item.find('p',class_="star-rating")['class'][1]

  row={
      'Book_Name':book_name,
      "Book_Link":book_link,
      'price':price,
      'stock':stock,
      'rating':rating
  }
  Book_info.append(row)
```

```python
df=pd.DataFrame(Book_info)
df
```

```python
dict1={
    'one':1,'two':2,'three':3,'four':4,'five':5
}
def f(a):
  return dict1[a.lower()]
df['rating']=df['rating'].apply(f)
df
```

```python
df.to_csv("Book_info.csv")
```

# All page data scrap

```python
Book_info_all=[]
dict1={
    'one':1,'two':2,'three':3,'four':4,'five':5
}
for i in range(1,51):
  url_all=f"https://books.toscrape.com/catalogue/page-{i}.html"
  r=requests.get(url_all)
  soup=BeautifulSoup(r.content,'html.parser')
  book=soup.find_all('li',class_="col-xs-6 col-sm-4 col-md-3 col-lg-3")
  for item in book:
    name_tag=item.find('h3').find('a')
    book_name=name_tag.text
    book_link=name_tag['href']
    price=item.find('p',class_="price_color").text
    stock=item.find('p',class_="instock availability").text.strip()
    rating=item.find('p',class_="star-rating")['class'][1]

    row={
        'Book_Name':book_name,
        "Book_Link":book_link,
        'price':price,
        'stock':stock,
        'rating':dict1[rating.lower()]
    }
    Book_info_all.append(row)
```

```python
df_all=pd.DataFrame(Book_info_all)
df_all
```

```python
df_all.to_csv("Book_info_all.csv")
```
